```python
import os
import requests
import time

def download_hashes(apikey, hashes, save_dir):
    """
    Downloads files from VirusShare based on a list of hashes and saves them to the specified directory.
    Retries if the response status is not 200.

    :param apikey: Your VirusShare API key.
    :param hashes: List of hashes to download.
    :param save_dir: Directory where files will be saved.
    """
    # Ensure the save directory exists
    os.makedirs(save_dir, exist_ok=True)

    base_url = "https://virusshare.com/apiv2/download"
    
    for hash_value in hashes:
        success = False
        while not success:
            url = f"{base_url}?apikey={apikey}&hash={hash_value}"
            try:
                print(f"Downloading hash: {hash_value}")
                response = requests.get(url, stream=True)

                if response.status_code == 200:
                    file_path = os.path.join(save_dir, f"{hash_value}.zip")
                    with open(file_path, "wb") as file:
                        for chunk in response.iter_content(chunk_size=8192):
                            file.write(chunk)
                    print(f"Saved: {file_path}")
                    success = True
                else:
                    print(f"Failed to download {hash_value}: HTTP {response.status_code}")
                    print(f"Response: {response.text}")
                    print("Retrying in 1 minute...")
                    time.sleep(60)  # Wait for 1 minute before retrying

            except Exception as e:
                print(f"Error downloading {hash_value}: {e}")
                print("Retrying in 1 minute...")
                time.sleep(60)  # Wait for 1 minute before retrying
```
